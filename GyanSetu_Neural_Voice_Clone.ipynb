```python
# Step 2: Fetch & Verify Simran Ailani's Clean Reference Voice Sample
import os
import urllib.request
import subprocess

os.makedirs("reference", exist_ok=True)
os.makedirs("raw_wavs", exist_ok=True)
os.makedirs("gyansetu_dub_masters", exist_ok=True)

ref_sample_path = "reference/simran_reference_sample.wav"

# Primary URL from GitHub repository
github_url = "https://raw.githubusercontent.com/tanmaythakre2006-cmd/GYANSETU/main/assets/audio_reference/simran_reference_sample.wav"
# Fallback URL from live Vercel deployment
vercel_url = "https://gyansetu-five.vercel.app/assets/audio/ep1-can-chemistry-explain-consciousness.mp3"

downloaded = False
try:
    print("📥 Fetching Simran Ailani's reference voice sample from repository...")
    urllib.request.urlretrieve(github_url, ref_sample_path)
    if os.path.exists(ref_sample_path) and os.path.getsize(ref_sample_path) > 10000:
        downloaded = True
        print(f"✅ Primary voice sample verified ({os.path.getsize(ref_sample_path)} bytes)")
except Exception as e:
    print(f"Notice on primary fetch: {e}")

if not downloaded:
    print("📥 Fetching master audio stream from live GyanSetu production server...")
    temp_mp3 = "reference/temp_stream.mp3"
    urllib.request.urlretrieve(vercel_url, temp_mp3)
    # Extract 12-second speech sample (00:00:04 to 00:00:16) with FFmpeg
    subprocess.run([
        "ffmpeg", "-y", "-ss", "00:00:04", "-t", "12",
        "-i", temp_mp3, "-ar", "24000", "-ac", "1", "-c:a", "pcm_s16le",
        ref_sample_path
    ], check=True)
    print("✅ Reference sample sliced & formatted into 24kHz mono PCM WAV!")

print(f"🎯 Reference Sample Ready: {ref_sample_path}")
```
